# Make an audiobook from an AO3 fic

This turns an EPUB you downloaded from AO3 into an audiobook read by a natural voice (Kokoro), using a free Google computer. Then you download one file and play it in **BookPlayer** or **Apple Books**: it works offline, keeps playing with the screen locked, and remembers your place.

**How:** tap the ▶ next to each step below, one at a time, and wait for each to finish before the next. On a phone, the ▶ appears when you tap a step.

Please keep the audio for your own listening; sharing it needs the author's permission.

```python
#@title 1. Get ready (takes a couple of minutes)
!apt-get -qq -y install espeak-ng > /dev/null 2>&1
!pip install -q "kokoro>=0.9.4" soundfile "ao3-reader @ git+https://github.com/wjiamin/text-reader"
from ao3reader.colab import graphics_chip
if graphics_chip():
    print("Ready, with a graphics chip ✓")
else:
    print("Ready, but this computer has no graphics chip, so it will be much slower.\n"
          "To fix it: Runtime → Change runtime type → T4 GPU → Save, then run this step again.")
```

```python
#@title 2. Choose the EPUB you downloaded from AO3
from google.colab import files
uploaded = files.upload()
epub = next((name for name in uploaded if name.lower().endswith(".epub")), None)
print(f"Got {epub}" if epub else "That wasn't an EPUB. On AO3, tap Download, then EPUB, and try again.")
```

```python
#@title 3. Make the audiobook
voice = "af_heart"  #@param ["af_heart", "af_bella", "af_nicole", "am_michael", "am_fenrir", "bf_emma", "bm_george", "bm_fable"]
read_summaries = True  #@param {type:"boolean"}
read_authors_notes = False  #@param {type:"boolean"}
from ao3reader.colab import make
audiobook = make(epub, voice=voice, summary=read_summaries, notes=read_authors_notes)
```

```python
#@title 4. Download the audiobook
files.download(str(audiobook))
```

```python
#@title Or: save it to your Google Drive instead (better for very long fics)
from google.colab import drive
import shutil
drive.mount("/content/drive")
shutil.copy(audiobook, "/content/drive/MyDrive/")
print("Saved to your Google Drive as", audiobook.name, "- open the Google Drive app on your phone to get it.")
```

**Then, on an iPhone:** open the downloaded file in the Files app and share it to BookPlayer (free) or Books. **On Android:** open it with an audiobook app such as Voice.

Voices: *af_* American female, *am_* American male, *bf_* British female, *bm_* British male. **af_heart** is the best-rated.
